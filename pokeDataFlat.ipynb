#### Pokemon Data JSON to table

```python
# reads json data file
df = spark.read.option("multiline", "true").json('Files/pokedata/poke1.json')
#creats a temp view to store data and use the explode function to get all the key & value pairs from json
df.createOrReplaceTempView("df_view")
```

```python
# create a dbo schema in tables section
spark.sql("CREATE SCHEMA IF NOT EXISTS pokedata")
```

```python
#fetches data from df_view
poke1 = spark.sql("select result.name as name, result.url as url from (select explode(results) as result from df_view) as a")
#saves data as delta file to table folder
poke1.write.mode("overwrite").format("delta").saveAsTable("pokedata.poke1")
```
